##  Loop/ciclos "For"

```python
numeros = [1, 2, 3, 4, 5, 6]

for numeros in numeros:
    print(numeros)
```

## Loop/ciclo "while"

```python
contador = 5

while contador > 0:
    print(contador)
    contador = contador - 1
```
